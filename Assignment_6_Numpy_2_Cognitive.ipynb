```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a) Add 2°C to every reading using vectorization
corrected = temperature + 2
print("Corrected temperatures:", corrected)

# b) Convert corrected temperatures from Celsius to Fahrenheit
fahrenheit = (9/5) * corrected + 32
print("Temperature in Fahrenheit:", fahrenheit)

# c) Identify readings greater than 32°C using Boolean indexing
above_32 = corrected[corrected > 32]
print("Readings greater than 32°C:", above_32)

# d) Count readings greater than 32°C
count = np.sum(corrected > 32)
print("Number of readings greater than 32°C:", count)

# e) Explanation
print("\nExplanation:")
print("Vectorization performs operations on an entire array at once,")
print("making the code faster and simpler than using a for loop.")
print("Boolean indexing helps select only the values that satisfy")
print("a given condition without manually checking each element.")
```

Q2)

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a) Total steps
print("Total steps:", np.sum(steps))

# b) Mean steps
print("Mean steps:", np.mean(steps))

# c) Maximum and minimum
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))

# d) Total steps for each day
print("Total steps for each day:", np.sum(steps, axis=0))

# e) Total steps for each user
print("Total steps for each user:", np.sum(steps, axis=1))

# f) Position of maximum steps
position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum steps:", position)
```

```python
# a) Create original array
original = np.array([1, 2, 3, 4, 5, 6])

# b) Slice
subset = original[1:4]

# c) Modify subset
subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

# d) Create a copy and modify it
copy_array = original[1:4].copy()
copy_array[0] = 500

print("Original after modifying copy:", original)
print("Copied array:", copy_array)

# e) Create 1 to 12 and reshape into 3 x 4
matrix = np.arange(1, 13).reshape(3, 4)

print("\n3 x 4 Matrix:")
print(matrix)

# f) Extract required elements
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

# g) Flatten and ravel
flat = matrix.flatten()
rav = matrix.ravel()

print("\nUsing flatten():", flat)
print("Using ravel():", rav)

# h) Modify ravel
rav[0] = 100
print("\nAfter modifying ravel:")
print("Ravel:", rav)
print("Original matrix:")
print(matrix)

# i) Modify flatten
flat[1] = 200
print("\nAfter modifying flatten:")
print("Flatten:", flat)
print("Original matrix:")
print(matrix)

# j) Display properties
print("\nShape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a) Shape and dimensions
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

# b) Transpose
print("\nTranspose of X:")
print(X.T)

# c) Matrix product X.T @ X
XtX = X.T @ X
print("\nX.T @ X:")
print(XtX)

# d) Inverse
inverse = np.linalg.inv(XtX)
print("\nInverse:")
print(inverse)

# e) OLS coefficients
beta = inverse @ X.T @ y
print("\nOLS coefficients:")
print(beta)

# f) Coefficients represent the effect of
# sleep, activity and stress on the predicted score.

# g) Predict assistance score
new_user = np.array([5, 40, 7])
prediction = new_user @ beta

print("\nPredicted assistance score:", prediction)
```

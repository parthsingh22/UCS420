```python
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = """Tennis is my favorite sport because it combines skill, fitness,
and strategy. I enjoy playing tennis because every point challenges
me to think and react quickly. Practicing regularly helps improve
my footwork, accuracy, and confidence. I also like watching
professional players and learning new techniques from them.
For me, tennis is a great way to stay active and enjoy my free time."""

# 1. Convert text to lowercase and remove punctuation
clean_text = re.sub(r'[^\w\s]', '', text.lower())

print("Cleaned Text:")
print(clean_text)

# 2. Tokenize into words and sentences
words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("\nWord Tokens:")
print(words)

print("\nSentence Tokens:")
print(sentences)

# 3. Compare split() and word_tokenize()
split_words = clean_text.split()
nltk_words = word_tokenize(clean_text)

print("\nUsing split():")
print(split_words)

print("\nUsing word_tokenize():")
print(nltk_words)

print("\nComparison:")
print("split() divides text using whitespace.")
print("word_tokenize() uses NLP rules to tokenize text.")

# 4. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]

print("\nWords after removing stopwords:")
print(filtered_words)

# 5. Display word frequency distribution
frequency = FreqDist(filtered_words)

print("\nWord Frequency Distribution:")
for word, count in frequency.most_common():
    print(word, ":", count)

frequency.plot(10, title="Word Frequency Distribution")
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

text = """Tennis is my favorite sport because it combines skill, fitness,
and strategy. I enjoy playing tennis because every point challenges
me to think and react quickly. Practicing regularly helps improve
my footwork, accuracy, and confidence. I also like watching
professional players and learning new techniques from them.
For me, tennis is a great way to stay active and enjoy my free time."""

# 1. Extract alphabetic words using regular expressions
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

print("Extracted Words:")
print(words)

# 2. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]

print("\nWords after removing stopwords:")
print(filtered_words)

# 3. Perform stemming
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("\nStemmed Words:")
print(stemmed_words)

# 4. Perform lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\nLemmatized Words:")
print(lemmatized_words)

# 5. Compare the two methods
print("\nComparison:")
print("Stemming removes word endings and may produce non-dictionary words.")
print("Lemmatization tries to return meaningful dictionary words.")

```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

texts = [
    "Artificial intelligence helps machines learn and solve problems.",
    "Machine learning allows computers to learn from data.",
    "Artificial intelligence and machine learning are used in modern technology."
]

# 1. Bag of Words representation
count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)

print("Bag of Words:")
print(count_vectorizer.get_feature_names_out())
print(bow_matrix.toarray())

# 2. TF/IDF representation
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

feature_names = tfidf_vectorizer.get_feature_names_out()

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())

# 3. Display top 3 keywords from each text
print("\nTop 3 Keywords from Each Text:")

for i in range(len(texts)):
    scores = tfidf_matrix[i].toarray().flatten()
    top_indices = scores.argsort()[-3:][::-1]

    print(f"\nText {i + 1}:")
    for index in top_indices:
        print(feature_names[index], ":", round(scores[index], 3))
```

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text1 = """
Artificial intelligence enables computers to learn from data and make decisions.
It is used in healthcare, education, automation, and recommendation systems.
AI helps solve complex problems and improves decision making.
"""

text2 = """
Blockchain is a distributed digital ledger that stores transactions securely.
It is used in cryptocurrency, supply chains, and digital identity systems.
Blockchain provides transparency and data security.
"""

# 1. Preprocessing and tokenization
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    return text.split()

words1 = preprocess(text1)
words2 = preprocess(text2)

print("Tokens of Text 1:")
print(words1)

print("\nTokens of Text 2:")
print(words2)

# 2a. Jaccard Similarity
set1 = set(words1)
set2 = set(words2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard = len(intersection) / len(union)

print("\nCommon Words:")
print(intersection)

print("Jaccard Similarity:", round(jaccard, 4))

# 2b. Cosine Similarity using TF-IDF
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform([text1, text2])

cosine = cosine_similarity(tfidf_matrix[0], tfidf_matrix[1])[0][0]

print("Cosine Similarity:", round(cosine, 4))

# 2c. Comparison
print("\nComparison:")
print("Jaccard similarity compares the common unique words.")
print("Cosine similarity compares the TF-IDF weighted word vectors.")
```

```python
from textblob import TextBlob
import matplotlib.pyplot as plt
from wordcloud import WordCloud

reviews = [
    "This laptop is amazing, fast, and easy to use. I really love it.",
    "The headphones have terrible sound quality and the battery is disappointing.",
    "The product is okay. It works as expected but nothing special."
]

positive_reviews = []

# 1. Find polarity and subjectivity
# 2. Classify each review
print("Sentiment Analysis Results:\n")

for review in reviews:
    analysis = TextBlob(review)

    polarity = analysis.sentiment.polarity
    subjectivity = analysis.sentiment.subjectivity

    if polarity > 0:
        sentiment = "Positive"
        positive_reviews.append(review)
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("Review:", review)
    print("Polarity:", round(polarity, 2))
    print("Subjectivity:", round(subjectivity, 2))
    print("Sentiment:", sentiment)
    print("-" * 50)

# 3. Create a word cloud from positive reviews
positive_text = " ".join(positive_reviews)

if positive_text:
    wordcloud = WordCloud(
        width=800,
        height=400,
        background_color="white",
        colormap="viridis"
    ).generate(positive_text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Word Cloud of Positive Reviews")
    plt.show()
else:
    print("No positive reviews found.")
```

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# 1. Training paragraph
text = """
Technology is changing the way people learn and communicate.
Artificial intelligence helps computers understand information
and perform useful tasks. Machine learning allows systems to
learn from examples and improve their performance.
Students can use technology to explore new ideas, practice skills,
and solve problems. Online learning makes education more accessible.
However, technology should be used responsibly and thoughtfully.
The future of technology depends on creativity, knowledge,
and the decisions people make. Learning new technology can
help us prepare for future challenges and opportunities.
"""

# 2. Tokenize the text
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Total unique words:", total_words)

# 3. Create input sequences
sequences = []

for line in text.split("."):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        sequence = token_list[:i + 1]
        sequences.append(sequence)

# Pad the sequences
max_sequence_length = max(len(seq) for seq in sequences)

sequences = pad_sequences(
    sequences,
    maxlen=max_sequence_length,
    padding="pre"
)

# 4. Separate input and output
X = sequences[:, :-1]
y = sequences[:, -1]

print("Input shape:", X.shape)
print("Output shape:", y.shape)

# 5. Build the LSTM model
model = Sequential([
    Embedding(total_words, 32, input_length=X.shape[1]),
    LSTM(64),
    Dense(total_words, activation="softmax")
])

model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

# 6. Train the model
model.fit(X, y, epochs=100, verbose=1)

# 7. Generate text using a seed word
def generate_text(seed_text, next_words):
    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=X.shape[1],
            padding="pre"
        )

        predicted_probs = model.predict(token_list, verbose=0)[0]
        predicted_index = np.argmax(predicted_probs)

        output_word = ""

        for word, index in tokenizer.word_index.items():
            if index == predicted_index:
                output_word = word
                break

        if output_word:
            seed_text += " " + output_word

    return seed_text

# 8. Generate 2-3 lines of text
print("\nGenerated Text:\n")

print(generate_text("technology", 10))
print(generate_text("learning", 10))
print(generate_text("artificial", 10))
```

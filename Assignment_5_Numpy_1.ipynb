ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])
values, counts = np.unique(x, return_counts=True)
max_count = max(counts)

for i in range(len(values)):
    if counts[i] == max_count:
        print("Most frequent value:", values[i])
        print("Indices:", np.where(x == values[i])[0])
```

```python
y = np.array([1,1,1,2,3,4,2,4,3,3])
values, counts = np.unique(y, return_counts=True)
max_count = max(counts)

for i in range(len(values)):
    if counts[i] == max_count:
        print("Most frequent value:", values[i])
        print("Indices:", np.where(y == values[i])[0])
```

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly
spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array,
shape, total elements, the data type of each element and total number of bytes consumed
by the array. Find the transpose of this array using reshape() attribute. Can we do the same
with T attribute?

```python
parth = np.linspace(10, 100, 25)
print("Array:")
print(parth)


print("Dimensions:", parth.ndim)
print("Shape:", parth.shape)
print("Total elements:", parth.size)
print("Data type:", parth.dtype)
print("Total bytes consumed:", parth.nbytes)


reshaped_parth = parth.reshape(5, 5)
print("\nReshaped array:")
print(reshaped_parth)


print("\nTranspose using reshape:")
print(reshaped_parth.reshape(5, 5).T)


print("\nTranspose using T:")
print(reshaped_parth.T)
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>>
with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean,
median, max, min, unique elements. Reshape the array to four rows and three columns and
name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three
columns and name it as resized_ ucs420_<your_name>>.

```python
ucs420_parth = np.array([[10, 20, 30, 40],
                         [50, 60, 70, 80],
                         [90, 15, 20, 35]])

print("Original array:")
print(ucs420_parth)


print("\nMean:", np.mean(ucs420_parth))
print("Median:", np.median(ucs420_parth))
print("Maximum:", np.max(ucs420_parth))
print("Minimum:", np.min(ucs420_parth))


print("Unique elements:", np.unique(ucs420_parth))


reshaped_ucs420_parth = ucs420_parth.reshape(4, 3)
print("\nReshaped array:")
print(reshaped_ucs420_parth)


resized_ucs420_parth = np.resize(ucs420_parth, (2, 3))
print("\nResized array:")
print(resized_ucs420_parth)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

Q1. Ask the user to enter a value (e.g., M) for a mathematical function.

 Generate x values from -10 to 10 using np.linspace().

 Compute y values for:

o y= M⋅x2

o y=M⋅sin(x)

 Plot both functions in a single figure:

o Use different colors and line styles.

o Add legend, grid, and title.

```python
M = float(input("Enter the value of M: "))

x = np.linspace(-10, 10, 100)

y1 = M * x**2
y2 = M * np.sin(x)

plt.plot(x, y1, 'r-', label='y = M*x²')
plt.plot(x, y2, 'b--', label='y = M*sin(x)')

plt.title('Plots of y = M*x² and y = M*sin(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)

plt.show()
```

Q.2 Create a dataset of five subjects and scores.

 Convert it into a Pandas DataFrame.

 Plot the scores using a Seaborn bar plot with:

o Different colors for each bar.

o Annotations on top of each bar.

o Title, axis labels, and grid.

```python
data = {
    'Subject': ['Math', 'Physics', 'Chemistry', 'English', 'Computer'],
    'Score': [85, 78, 92, 88, 95]
}
df = pd.DataFrame(data)
print(df)


plt.figure(figsize=(8, 5))
ax = sns.barplot(x='Subject', y='Score', data=df, palette='Set2')


for i, score in enumerate(df['Score']):
    ax.text(i, score + 1, str(score), ha='center')

plt.title('Scores of Five Subjects')
plt.xlabel('Subjects')
plt.ylabel('Scores')


plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

Q3. Write a Python script to select your roll number as seed for NumPy and
then generate a dataset of 50 values using np.random.randn().

 Create a 2x2 subplot layout displaying:

o Line plot showing cumulative sum.

o Scatter plot with random noise.

 Customize with titles, axis labels, and grids.

```python
np.random.seed(1024170086)

data = np.random.randn(50)

x = np.arange(1, 51)

cumulative_sum = np.cumsum(data)

# Create a 2x2 subplot layout
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# 1) Line plot showing cumulative sum
axes[0, 0].plot(x, cumulative_sum, color='blue')
axes[0, 0].set_title('Cumulative Sum')
axes[0, 0].set_xlabel('Index')
axes[0, 0].set_ylabel('Cumulative Value')
axes[0, 0].grid(True)

# 2) Scatter plot showing random noise
axes[0, 1].scatter(x, data, color='red')
axes[0, 1].set_title('Scatter Plot of Random Noise')
axes[0, 1].set_xlabel('Index')
axes[0, 1].set_ylabel('Random Value')
axes[0, 1].grid(True)

# 3) Histogram of random values
axes[1, 0].hist(data, bins=10, color='green', edgecolor='black')
axes[1, 0].set_title('Histogram of Random Values')
axes[1, 0].set_xlabel('Value')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].grid(True)

# 4) Line plot of original random values
axes[1, 1].plot(x, data, color='purple', marker='o', markersize=3)
axes[1, 1].set_title('Random Values')
axes[1, 1].set_xlabel('Index')
axes[1, 1].set_ylabel('Value')
axes[1, 1].grid(True)


plt.tight_layout()

plt.show()
```

Q.4 Download Data-set from the below link
https://github.com/AnjulaMehto/MCA/blob/main/company_sales_data.csv
Apply ‘seaborn’ library to do the following.

1. Read Total profit of all months and show it using a line plot.

2. Read all product sales data and show it using a multiline plot.

3. Plot bar chart for all the features/attributes.

```python
url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"
df = pd.read_csv(url)

print(df.head())

plt.figure(figsize=(10, 6))

sns.lineplot(data=df, x='month_number', y='total_profit', marker='o')

plt.title('Total Profit per Month')
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.grid(True)

plt.show()


plt.figure(figsize=(12, 6))

products = [
    'facecream',
    'facewash',
    'toothpaste',
    'bathingsoap',
    'shampoo',
    'moisturizer'
]

for product in products:
    sns.lineplot(data=df, x='month_number', y=product, marker='o', label=product)

plt.title('Monthly Sales of All Products')
plt.xlabel('Month Number')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)

plt.show()


features = [
    'facecream',
    'facewash',
    'toothpaste',
    'bathingsoap',
    'shampoo',
    'moisturizer',
    'total_units',
    'total_profit'
]

fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.flatten()

for i, feature in enumerate(features):
    sns.barplot(data=df, x='month_number', y=feature, color='skyblue', ax=axes[i])

    axes[i].set_title(f'Monthly {feature}')
    axes[i].set_xlabel('Month Number')
    axes[i].set_ylabel(feature)
    axes[i].grid(axis='y')


axes[8].axis('off')

plt.tight_layout()
plt.show()
```
